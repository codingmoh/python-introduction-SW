| Code cell | Python instructions sent to the kernel |
| Output area | Text or another result produced by the last execution |

## 3. Variables connect cells

A variable name refers to a value in the running kernel. Run the next two cells in order. The second cell uses the `price` created in the first.

```python
price = 12
quantity = 3
```

```python
total = price * quantity
print("Total:", total)
assert total == 36
```

The expected total is `36`. An `assert` checks a condition and raises `AssertionError` if it is false. The examples and solutions use assertions to make expectations executable.

**Try it:** change `price` to `15`, run its cell, and run the calculation cell again. The total becomes `45`, so the original assertion fails until you update its expectation. This is useful feedback: the check describes the earlier input.

Restore `price = 12` and the original assertion before continuing. If you change source text but do not execute the cell, Python still remembers the old value.

## 4. Errors are information

An error message identifies the error type and points to the operation that failed. Read the final line of the traceback first, then inspect the relevant line of your code.

This example deliberately attempts an invalid conversion. It catches the expected `ValueError` so that the whole notebook can continue running. Chapter 05 explains exception handling in detail.

```python
try:
    int("banana")
except ValueError as error:
    print(type(error).__name__ + ":", error)

print("The notebook can continue after the handled example.")
```

| Error | Common cause |
| --- | --- |
| `SyntaxError` | Python cannot parse the code, for example an unclosed quote |
| `NameError` | A name has not been defined in this kernel |
| `TypeError` | An operation receives an unsuitable kind of value |
| `ValueError` | The kind of value is accepted, but this particular value is invalid |
| `AssertionError` | A stated expectation is false |

A failed cell does not automatically undo assignments or other operations that happened earlier in the cell. After fixing an error, check the state you are relying on.

## 5. Restart and reproduce

Use **Kernel → Restart Kernel and Run All Cells** to check the notebook from a clean state. The exact menu wording can vary slightly between notebook applications.

A restart discards variables and imports. Run All executes the cells from top to bottom. Every supplied notebook is designed to work this way, independently of the other notebooks.

A spinning or busy kernel may be working or stuck in a loop. Use **Interrupt Kernel** to stop the running computation. Restart if you need a clean state afterwards.

## 6. Check the Python environment

The kernel must use the repository's environment. The command `uv run --locked jupyter lab` launches Jupyter in it. In VS Code, choose the `.venv` Python interpreter as the kernel.

The next cell reports the running Python minor version and whether it is in a virtual environment. The course configuration selects Python 3.13.

```python
import sys

print("Python:", ".".join(str(part) for part in sys.version_info[:3]))
print("Virtual environment:", sys.prefix != sys.base_prefix)
assert sys.version_info[:2] == (3, 13), "Select this repository's Python 3.13 kernel."
```
