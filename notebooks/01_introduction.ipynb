## 1. Readability and a first program

Python emphasizes readable code. The **Zen of Python** is a set of design aphorisms,
not rules that replace judgment. Its themes include readable names, explicit behavior,
simplicity, visible errors, and useful namespaces. For example, naming a result
`total_price` communicates more than naming it `x`.

Unlike a typical introductory Java or C++ program, a Python script can start with
one call to `print`; a class and a `main` function are not required. Semicolons are
unnecessary at line endings. `print` writes text to the output area. A notebook also
displays the value of the **last expression** in a code cell; a script does not.


````python
print("Hello world!")
unit_price = 4
quantity = 3
total_price = unit_price * quantity
print("Total:", total_price)
total_price  # A notebook additionally displays 12 as the last expression.
````

## 2. Objects have types; names refer to objects

An assignment such as `x = 2` binds the name `x` to an integer object. Reassigning
`x = "Hello"` binds that name to a string object. The integer did not turn into a
string. You do not have to declare a fixed type for the name. This is **dynamic typing**.

`type(value)` returns the type of an object. `None` represents the absence of a useful
value; it is different from zero or the empty string. Type objects are themselves
objects: `type(int)` is `type`. Choose descriptive `snake_case` names, and avoid names
such as `str`, `list`, and `print` that would hide useful built-ins.


````python
x = 2
print(x * 7, type(x).__name__)   # 14 int
x = "Hello, I'm "
print(x + "Python!", type(x).__name__)
print(type(1), type("Hello"), type(None))
print(type(int), type(type))
missing_value = None
assert missing_value is None
````

## 3. Numbers and arithmetic

`int` represents integers of arbitrary precision, subject to available memory. `float`
usually uses binary floating-point and therefore approximates many decimal fractions.
Python also has built-in `complex` numbers; the slides' claim that there are only two
numeric types is incomplete. `bool` is a subclass of `int`, used for logical results.

Operators follow precedence: `**` before multiplication/division, then addition and
subtraction. Parentheses make the intended grouping explicit.

| Expression | Meaning | Example result |
|---|---|---|
| `a / b` | true division | `5 / 2` → `2.5` |
| `a // b` | floor division | `7 // 3` → `2` |
| `a % b` | remainder | `7 % 3` → `1` |
| `a ** b` | exponentiation | `2 ** 4` → `16` |

Floor division rounds **down**, including for negative results. The stop value in
later `range` examples is unrelated to division and is always excluded.


````python
print(1 + 1, 8 - 1, 10 * 2)  # 2 7 20
print(9 / 3, 5 / 2, 7 / 1.4)  # 3.0 2.5 5.0
print(7 // 3, 7 % 3, 2 ** 4)  # 2 1 16
print(-7 // 3, -7 % 3)        # -3 2
assert -7 == (-7 // 3) * 3 + (-7 % 3)
print(2 + 3 * 4, (2 + 3) * 4)  # 14 20
print(10 ** 30)  # Integers do not overflow at a fixed 32-bit boundary.
print((1 + 2j) + (3 - 1j))     # (4+1j)
````

````python
import math

print(0.1 + 0.2)  # 0.30000000000000004 on the usual float implementation
print(0.1 + 0.2 == 0.3)  # False: exact equality compares stored values.
assert math.isclose(0.1 + 0.2, 0.3)
try:
    print(5 / 0)
except ZeroDivisionError as error:
    print("Expected error:", type(error).__name__)
````

## 4. Booleans, comparisons, and short-circuit evaluation

`True` and `False` have capital initials. `==` compares values; `=` assigns a name.
Use `!=`, `<`, `<=`, `>`, and `>=` for other comparisons. A chained comparison
`1 < value < 10` means that both bounds hold.

`not`, `and`, and `or` combine conditions. `and` evaluates its second operand only if
the first is truthy; `or` evaluates it only if the first is falsy. This can prevent an
invalid operation. With non-Boolean operands, `and` and `or` return an **operand**, not
necessarily `True` or `False`. `bool(...)` explicitly requests a Boolean.


````python
print(not True, True and False, True or False)  # False False True
print(1 == 1, 2 * 3 != 5, 2 >= 0)             # True True True
print(1 < 2 < 3, 1 < 2 >= 3)                 # True False
print(isinstance(True, int), True == 1, False == 0)
denominator = 0
safe_test = denominator != 0 and 10 / denominator > 2
print(safe_test)  # False: division was skipped.
print("" or "anonymous", "ready" and 7)  # anonymous 7
````

## 5. Strings: Unicode text and indexing

Single and double quotes both create `str` objects. Python has no separate character
type: indexing a string returns a string of length one. Strings are Unicode text.
`+` concatenates strings and `*` repeats them. Strings are immutable, so an individual
character cannot be replaced in place.

Indexes start at zero. For `"Arthur"`, valid positive indexes are 0 through 5; index 6
is just beyond the end. Negative indexes count from the right: −1 is the final
character and −6 is the first. `-0` equals `0`, not the end. Unicode code points are
not always the same as what a reader sees as one character (for example, combining
marks); these lessons use text where that distinction does not affect slicing.


````python
greeting = 'Hello'
group = "wørld"
print(greeting + " " + group + "!")
word = "Arthur"
print(word[0], word[1], word[4])    # A r u
print(word[-1], word[-2], word[-4], word[-6])  # r u t A
assert type(word[0]) is str
try:
    print(word[6])
except IndexError:
    print("Expected IndexError: index 6 is outside this six-letter string.")
try:
    word[0] = "a"
except TypeError:
    print("Expected TypeError: strings cannot be changed in place.")
````

## 6. Slices select a subsequence

`sequence[start:stop:step]` includes `start` and excludes `stop`. Omitted bounds use
the appropriate beginning/end for the direction of the step. A negative step walks
backwards. A slice can be empty, and out-of-range slice bounds are clipped instead
of raising `IndexError`. A step of zero is invalid.

For `"Arthur"`, `word[1:5:2]` visits indexes 1 and 3, producing `"rh"`.
`word[4::-2]` visits indexes 4, 2, and 0, producing `"utA"`.


````python
word = "Arthur"
print(word[0:2], word[3:6], word[1:4])  # Ar hur rth
print(word[:2], word[3:])              # Ar hur
print(word[1:5:2], word[4::-2], word[::-1])  # rh utA ruhtrA
print(repr(word[3:3]), word[:100])      # '' Arthur
assert word[-0] == "A"
try:
    print(word[::0])
except ValueError:
    print("Expected ValueError: a slice step cannot be zero.")
````

## 7. Convert values explicitly

`str`, `int`, and `float` construct converted values when the input is valid. Conversion
is not automatic for every operation: adding an integer to a string is ambiguous.
`int("2.5")` fails because that text is not an integer literal; `int(2.9)` truncates
towards zero, so it is not a general rounding function. Keep the original text if
you need to report invalid input later.


````python
print(str(42), int("42"), float("2.5"), float("1"))
print(int(2.9), int(-2.9))  # 2 -2
for raw in ["42", " 7 ", "2.5", "unknown"]:
    try:
        number = int(raw)
        print(repr(raw), "->", number)
    except ValueError:
        print(repr(raw), "is not valid integer text")
try:
    print("Age: " + 20)
except TypeError:
    print("Expected TypeError; use str(20) or a formatted string.")
print(f"Age: {20}")
````

## 8. Lists are ordered, mutable collections

Square brackets construct a list and commas separate its elements. A list can hold
values of different types and may contain other lists. In practice, a consistent
kind of element often makes a program easier to reason about.

`append` adds one item at the end and returns `None`. Assigning its result back to the
list name loses that name's reference to the list. Indexing and slicing follow the
same rules as for strings, but list items can be replaced. A slice creates a new
outer list; references to nested objects may still be shared (chapter 02).


````python
empty = []
letters = ["a", "b", "c", "d"]
numbers = [2, 3, 5]
mixed = [4, 5, "seconds"]
append_result = numbers.append(7)
numbers.append(11)
print(numbers, append_result)  # [2, 3, 5, 7, 11] None
print(numbers[0], numbers[-1], letters[:3], numbers[1:-1])
numbers[0] = 13
print(numbers)                # [13, 3, 5, 7, 11]
assert append_result is None
````

````python
letters = ["a", "b", "c", "d"]
numbers = [2, 3, 5, 7, 11]
nested = [letters, numbers]
print(nested[0], nested[0][1], nested[1][2:])
print(len([]), len("python"), len([4, 5, "seconds"]))  # 0 6 3
print(0 in [], "y" in "python", "minutes" in [4, 5, "seconds"])
print("seconds" not in numbers)  # True
````

## 10. Decisions and indentation

An `if` runs a block when its condition is truthy. An `elif` tries another condition
only if earlier conditions failed. An optional `else` handles the remaining cases.
Python uses a colon and indentation to delimit blocks: use **four spaces per level**,
keep sibling statements aligned, and do not mix tabs with spaces. Conditions do not
need surrounding parentheses.

A chain selects at most one branch. Separate `if` statements can select several.
Order overlapping conditions from specific to general. Modern Python also provides
`match`/`case` for structural pattern matching (since 3.10); the older slides predate
it. `if`/`elif` remains appropriate for the comparisons in this chapter.


````python
temperature = 18
if temperature < 0:
    print("freezing")
elif temperature < 20:
    print("cool")
else:
    print("warm")
print("This statement is outside the conditional.")

word = "level"
if word == word[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
# This simple palindrome check is case-sensitive and includes spaces/punctuation.
````

## 11. Truthy and falsy values

Common falsy values are `None`, `False`, numeric zero, and empty built-in collections
such as `""` and `[]`. Nonempty strings, including `"False"` and `"0"`, are truthy.
Use `if data:` to ask whether a collection has contents, or `if not data:` for emptiness.
`if data == True` asks a different question. User-defined classes can customize truth
behavior; the rule is not simply that every other object must be truthy.

Distinguish an absent value from a valid zero: use `value is None` when absence is
what matters. Otherwise `if not value` would group the two cases together.


````python
for value in [None, False, 0, 0.0, "", [], "False", "0", [0]]:
    print(repr(value), "->", bool(value))
data = []
if data:
    print("Items:", len(data))
else:
    print("There's no data!")
measurement = 0
assert not measurement
assert measurement is not None  # Zero is a present measurement.
````

## 12. Loops iterate over data

A `for` loop takes each item from an iterable in order. Strings yield characters and
lists yield their elements. You often do not need a numeric index. The indented body
runs once per item, and an empty iterable runs it zero times.

`range(stop)`, `range(start, stop)`, and `range(start, stop, step)` describe integers
without building a list of all of them. The stop is **excluded**. The sign of the step
must agree with the direction; a zero step raises `ValueError`.


````python
for letter in "cat":
    print(letter, end=" ")  # c a t
print()
total = 0
for amount in [4, 7, 2]:
    total += amount
print("Total:", total)  # 13
for bounds in [(3,), (5, 10), (2, 12, 3), (-7, -30, -5), (5, 2)]:
    print(bounds, "->", list(range(*bounds)))
# *bounds expands the tuple into arguments; chapter 04 explores unpacking.
````

## 13. Break, continue, and a bounded while loop

`break` exits the smallest enclosing loop. `continue` skips the rest of the current
iteration and proceeds to the next. A `while` loop repeats while a condition is
truthy; make sure something changes so that it can end. Use a `for` loop when iterating
through known data is a better fit.

A loop may have an `else` block: it runs when the loop finishes without `break`, even
if there were zero iterations. It does not mean "the loop condition was false on entry
only". The prime-number example below instead uses early `return` for clarity.


````python
visited = []
for number in range(10):
    if number == 6:
        break
    visited.append(number)
print(visited)  # [0, 1, 2, 3, 4, 5]

odd_numbers = []
for number in range(8):
    if number % 2 == 0:
        continue
    odd_numbers.append(number)
print(odd_numbers)  # [1, 3, 5, 7]

remaining = 3
while remaining > 0:
    print(remaining, end=" ")
    remaining -= 1
print("done")
````

````python
for candidate in [4, 8, 12]:
    if candidate % 2:
        print("Found an odd number:", candidate)
        break
else:
    print("No odd number found")

# break only exits the inner loop here.
pairs = []
for row in range(2):
    for column in range(3):
        if column == 1:
            break
        pairs.append((row, column))
assert pairs == [(0, 0), (1, 0)]
print(pairs)
````

## 14. Functions name reusable behavior

`def` creates a function. **Parameters** are the names in its definition; **arguments**
are the values supplied when calling it. The body runs when the function is called,
not when it is defined. `return` ends that call and gives a value to its caller.
Reaching the end, or writing `return` alone, returns `None`.

Printing and returning solve different problems: printed text is visible output;
a returned value can be used in further calculations. Functions do not require type
annotations. Later chapters cover defaults, keyword arguments, variable argument
lists, unpacking, lambdas, functions as values, and functional techniques.


````python
def area(width, height):
    """Return the area of a rectangle."""
    return width * height


def announce(text):
    print(text)


result = area(3, 4)
print(result, result + 1)   # 12 13
announcement_result = announce("Functions can produce output.")
assert announcement_result is None
print(area.__doc__)
````

## 15. A complete example: prime numbers

A prime is an integer **at least 2** whose only positive divisors are 1 and itself.
For `n < 2`, return `False` immediately. Omitting that check would incorrectly call
negative integers, 0, and 1 prime because their divisor loop is empty.

A factor pair cannot have both factors larger than the square root of `n`. Therefore,
we only need divisors up to and including `math.isqrt(n)`, the integer square root.
The `+ 1` below includes that boundary and catches squares such as 49. This function
expects integer input; conversion of external text belongs at the input boundary.


````python
from math import isqrt


def is_prime(n):
    """Return whether integer n is prime."""
    if n < 2:
        return False
    for divisor in range(2, isqrt(n) + 1):
        if n % divisor == 0:
            return False
    return True


for number in [-3, 0, 1, 2, 3, 4, 9, 29, 49]:
    print(number, "is prime:", is_prime(number))
assert not is_prime(1)
assert is_prime(2)
assert not is_prime(49)
````

````python
simulated_limit = "12"
limit = int(simulated_limit)
primes = []
for number in range(2, limit):
    if is_prime(number):
        primes.append(number)
        print(number, "is prime")
    else:
        print(number, "is not prime")
print("Primes below", limit, ":", primes)
assert primes == [2, 3, 5, 7, 11]
````
