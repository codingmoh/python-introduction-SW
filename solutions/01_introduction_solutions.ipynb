## Exercise 1 · Temperature conversion

Return the computed number so callers can reuse it. True division preserves a
fractional result. The checks include two familiar reference points, a negative
value, and a fractional input. An f-string inserts the result into readable output.


```python
def celsius_to_fahrenheit(celsius):
    """Convert a numeric Celsius temperature to Fahrenheit."""
    return celsius * 9 / 5 + 32


assert celsius_to_fahrenheit(0) == 32
assert celsius_to_fahrenheit(100) == 212
assert celsius_to_fahrenheit(-40) == -40
assert celsius_to_fahrenheit(12.5) == 54.5
print(f"12.5 °C is {celsius_to_fahrenheit(12.5):.1f} °F.")
```

## Exercise 2 · Slicing and exact palindromes

Reversal uses a step of −1. No normalization is performed, so case and punctuation
participate in equality. The empty string equals its reverse and therefore returns
`True` under the exercise's stated convention.


```python
word = "Arthur"
first_two = word[:2]
last_three = word[-3:]
reversed_word = word[::-1]
assert (first_two, last_three, reversed_word) == ("Ar", "hur", "ruhtrA")


def is_palindrome(text):
    """Return whether text exactly equals its reverse."""
    return text == text[::-1]


assert is_palindrome("level")
assert not is_palindrome("Level")
assert is_palindrome("")
assert is_palindrome("x")
assert is_palindrome("a a")
assert not is_palindrome("level!")
print(first_two, last_three, reversed_word)
```

## Exercise 3 · Mutating a list and slicing it

`append` changes the existing list. Item assignment replaces one element. A slice
creates a new outer list, so appending to `recent` does not append to `basket`.
The list elements here are immutable strings, so no nested mutable state is involved.


```python
basket = ["bread", "milk"]
basket.append("apples")
basket[1] = "oat drink"
recent = basket[-2:]
assert basket == ["bread", "oat drink", "apples"]
assert recent == ["oat drink", "apples"]
assert recent is not basket
recent.append("tea")
assert len(basket) == 3
assert recent == ["oat drink", "apples", "tea"]
assert "bread" in basket
assert "milk" not in basket
assert [][-2:] == []  # Slicing an empty list is safe.
print(basket, recent)
```

## Exercise 4 · Decisions at boundaries

The first true branch wins. If `celsius < 20` were first, it would also capture
negative temperatures and incorrectly classify them as cool. The `else` handles
20 and every larger value, so the upper boundary belongs to the warm category.


```python
def classify_temperature(celsius):
    """Return freezing, cool, or warm for the exercise's temperature bands."""
    if celsius < 0:  # Test the narrower, coldest category first.
        return "freezing"
    elif celsius < 20:
        return "cool"
    else:
        return "warm"


assert classify_temperature(-1) == "freezing"
assert classify_temperature(0) == "cool"
assert classify_temperature(19.9) == "cool"
assert classify_temperature(20) == "warm"
assert classify_temperature(100) == "warm"
print([classify_temperature(value) for value in [-1, 0, 19.9, 20]])
```

## Exercise 5 · Stop, skip, collect

Check the sentinel before the skip rule: zero is even, but it must stop processing.
The function returns a new list and leaves the input unchanged. Duplicate positive
odd values remain in their original order.


```python
def collect_positive_odds(values):
    """Collect positive odd integers before the first zero sentinel."""
    result = []
    for value in values:
        if value == 0:
            break
        if value < 0 or value % 2 == 0:
            continue
        result.append(value)
    return result


source = [3, -1, 4, 5, 0, 7]
assert collect_positive_odds(source) == [3, 5]
assert source == [3, -1, 4, 5, 0, 7]
assert collect_positive_odds([]) == []
assert collect_positive_odds([0, 3]) == []
assert collect_positive_odds([-3, 2, 4]) == []
assert collect_positive_odds([1, 1, 2, 3]) == [1, 1, 3]
print(collect_positive_odds(source))
```

## Exercise 6 · Prime numbers and an exclusive limit

The early guard rejects all integers below 2. Trial divisors include the integer
square root, catching perfect squares. For 2 the loop is empty after the guard,
which correctly returns `True`. A limit at or below 2 yields an empty outer range.


```python
from math import isqrt


def is_prime(n):
    """Return whether integer n is prime."""
    if n < 2:
        return False
    for divisor in range(2, isqrt(n) + 1):
        if n % divisor == 0:
            return False
    return True


def primes_below(limit):
    """Return all prime integers strictly below integer limit."""
    result = []
    for candidate in range(2, limit):
        if is_prime(candidate):
            result.append(candidate)
    return result


for nonprime in [-7, 0, 1, 4, 9, 25, 49, 121]:
    assert not is_prime(nonprime), nonprime
for prime in [2, 3, 5, 29, 97]:
    assert is_prime(prime), prime
assert primes_below(-10) == []
assert primes_below(2) == []
assert primes_below(3) == [2]
assert primes_below(12) == [2, 3, 5, 7, 11]
assert primes_below(11) == [2, 3, 5, 7]  # The limit is excluded.
print(primes_below(30))
```
