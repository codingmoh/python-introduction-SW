### 01-08 - Turn an algorithm into a function

**After PDF p. 51 · End of this PDF**  
`1_introduction_python.pdf` · Slide block 47-51

- Define is_prime(number) and use it to collect the primes below 12.
- Check 0, 1, 2, and 9. Explain why the small-number guard must come before the divisor loop.

```python
def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, number):
        if number % divisor == 0:
            return False
    return True

primes = []
for number in range(12):
    if is_prime(number):
        primes.append(number)
print(primes)
assert primes == [2, 3, 5, 7, 11]
assert not is_prime(0) and not is_prime(1)
assert is_prime(2) and not is_prime(9)
```

```python
note = "Grüße from Python!\n"

from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as directory:
    path = Path(directory) / "notes.txt"
    path.write_text(note, encoding="utf-8")
    try:
        with path.open(encoding="utf-8") as stream:
            content = stream.read()
            print(content)
            raise ValueError("An intentional demonstration")
    except ValueError as error:
        print(error)
    print("Closed:", stream.closed)
    assert stream.closed and content == note
```

### 04-03 - Defaults: one object or a fresh object?

**After PDF p. 21 · Resume at PDF p. 22**  
`4_functions.pdf` · Slide block 12-21

- Write a greeting with an optional prefix; call it positionally and by keyword. Catch a duplicate-argument error.
- Write a collector with a list default and call it twice. Contrast this with the None-and-fresh-list pattern.

```python
def greet(name, prefix="Hi"):
    return f"{prefix}, {name}!"


def collect_bad(value, values=[]):
    values.append(value)
    return values


def collect(value, values=None):
    if values is None:
        values = []
    values.append(value)
    return values


print(greet("Mira"), greet(prefix="Hello", name="Ali"))
try:
    greet("Mira", name="Ali")
except TypeError as error:
    print(type(error).__name__, "— name received two values")
print("Shared default, first call:", collect_bad("A"))
print("Shared default, second call:", collect_bad("B"))
print("Fresh defaults:", collect("A"), collect("B"))
assert collect("C") == ["C"]
```

### 05-04 - Inheritance, super, and MRO

```python
class Course:
    def __init__(self, title):
        self.title = title

    def describe(self):
        return self.title

class OnlineCourse(Course):
    def describe(self):
        return "online " + super().describe()

class RecordedCourse(Course):
    def describe(self):
        return "recorded " + super().describe()

class HybridCourse(OnlineCourse, RecordedCourse):
    pass

online = OnlineCourse("Python")
hybrid = HybridCourse("Python")
order = [cls.__name__ for cls in HybridCourse.__mro__]
print(online.describe(), hybrid.describe())
print(order)
assert online.describe() == "online Python"
assert hybrid.describe() == "online recorded Python"
assert order == ["HybridCourse", "OnlineCourse", "RecordedCourse", "Course", "object"]
```

- Make Point(x, y) support a readable representation, point addition, and iteration over its coordinates.
- Try print, repr, +, and tuple on your points. Inspect what happens when a point is added to an integer.

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):
        return f"Point({self.x!r}, {self.y!r})"

    def __add__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + other.y)

    def __iter__(self):
        return iter((self.x, self.y))

point = Point(3, 5)
total = point + Point(9, -2)
print(point, repr(total), tuple(total))
assert repr(point) == "Point(3, 5)"
assert tuple(total) == (12, 3)
try:
    point + 1
except TypeError as error:
    print(type(error).__name__)
else:
    raise AssertionError("Point + int must fail")
```

### 05-07 - Raise, custom exceptions, and else

**After PDF p. 85 · Resume at PDF p. 86**  
`5_oop.pdf` · Slide block 76-85

- Write reserve(places, available=2). Raise ValueError for nonpositive requests and a custom CapacityError when too many places are requested.
- Try requests of 1, 3, and 0. Handle each error specifically and report the remaining places only in else.

```python
class CapacityError(Exception):
    """A reservation exceeds the available places."""

def reserve(places, available=2):
    if places <= 0:
        raise ValueError("places must be positive")
    if places > available:
        raise CapacityError("not enough places")
    return available - places

outcomes = []
for requested in (1, 3, 0):
    try:
        remaining = reserve(requested)
    except CapacityError as error:
        outcomes.append("full")
        print(requested, type(error).__name__, str(error))
    except ValueError as error:
        outcomes.append("invalid")
        print(requested, type(error).__name__, str(error))
    else:
        outcomes.append(remaining)
        print(requested, "Remaining:", remaining)
assert outcomes == [1, "full", "invalid"]
```

### 05-08 - Cleanup with finally and with

**After PDF p. 89 · End of this PDF**

```python
from io import StringIO

def read_first(stream):
    try:
        return int(stream.readline())
    finally:
        stream.close()

for text in ("10\n", "invalid\n"):
    stream = StringIO(text)
    try:
        print("Read:", read_first(stream))
    except ValueError:
        print("Invalid integer")
    print("Closed:", stream.closed)
    assert stream.closed

with StringIO("20\n") as stream:
    value = int(stream.readline())
    print("Inside with:", value, stream.closed)
print("After with:", stream.closed)
assert value == 20 and stream.closed
```

### 06-04 - Iterator protocol and exhaustion

**After PDF p. 30 · Resume at PDF p. 31**  
`6_FP.pdf` · Slide block 25-30

- Build `Countdown(3)` with `__iter__` and `__next__`, ending with `StopIteration` at zero.
- Consume it with next and list, then try again. Compare the exhausted iterator with fresh iterators over a list.

```python
numbers = [1, 2, 3]


class Countdown:
    def __init__(self, start):
        self.remaining = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.remaining <= 0:
            raise StopIteration
        value = self.remaining
        self.remaining -= 1
        return value

countdown = Countdown(3)
assert iter(countdown) is countdown
print(next(countdown))
rest = list(countdown)
print(rest, list(countdown))
assert rest == [2, 1]
try:
    next(countdown)
except StopIteration:
    print("Exhausted")
else:
    raise AssertionError("Countdown must be exhausted")
print(list(iter(numbers)), list(iter(numbers)))
assert list(countdown) == []
assert list(iter(numbers)) == [1, 2, 3]
```

### 06-07 - Decorators that preserve behavior

**After PDF p. 49 · Resume at PDF p. 50**

- Build a debug decorator that prints arguments, forwards positional and keyword arguments, and returns the wrapped result.
- Decorate total(a, b, multiplier=1). Preserve its name and docstring with wraps, then inspect both.

```python
from functools import wraps

def debug(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print("Arguments:", args, kwargs)
        return function(*args, **kwargs)
    return wrapper

@debug
def total(a, b, multiplier=1):
    """Add two values and multiply the sum."""
    return (a + b) * multiplier

result = total(2, 3, multiplier=4)
print("Result:", result)
print(total.__name__, total.__doc__)
assert result == 20
assert total.__name__ == "total"
assert total.__doc__ == "Add two values and multiply the sum."
```

### 06-08 - Caching recursive results

**After PDF p. 50 · End of this PDF**  
`6_FP.pdf` · Slide block 50-50

- Cache recursive Fibonacci for small nonnegative integer indices. Inspect cache_info before and after repeating the same call.
- Use typed cache keys. Warm the cache with a valid integer keyword call, then verify that Boolean and float inputs still raise TypeError.

```python
from functools import lru_cache

@lru_cache(maxsize=None, typed=True)
def fibonacci(index):
    if type(index) is not int:
        raise TypeError("index must be an integer")
    if index < 0:
        raise ValueError("index must not be negative")
    if index < 2:
        return index
    return fibonacci(index - 1) + fibonacci(index - 2)

first = fibonacci(index=10)
before = fibonacci.cache_info()
second = fibonacci(index=10)
after = fibonacci.cache_info()
print(first, second)
print(before, after)
assert first == second == 55
assert after.hits == before.hits + 1
assert after.misses == before.misses
assert fibonacci(index=1) == 1
for invalid in (True, 1.0):
    try:
        fibonacci(index=invalid)
    except TypeError as error:
        print(repr(invalid), type(error).__name__)
    else:
        raise AssertionError("invalid type reached the cache")
```
